# OntoGPT + LinkML + BioPortal Agentic System
## Google ADK 2.0 Graph Workflow Prototype

**Full pipeline with:**
1. BioPortal ontology recommendation
2. **LLM-driven** LinkML / SPIRES template generation (few-shot from real OntoGPT templates)
3. Validation + **repair loop**
4. **OntoGPT SPIRES extraction agent** → structured output / RDF-ready
5. ADK 2.0-style `Workflow` / LoopAgent with conditional repair

> A SPIRES schema is a user-defined LinkML blueprint that guides an LLM to extract, nest, and ground entities & relationships from unstructured text (OntoGPT).

**Sample domain used in the demo:** Disease – Gene – Drug relations from a short clinical abstract.
